```python
df_dates = pd.DataFrame({'date': dates_2})
df_dates['date'] = pd.to_datetime(df_dates['date'])
```

```python
df_dates['ano'] = df_dates['date'].dt.year
df_dates['mes'] = df_dates['date'].dt.month
df_dates['dia'] = df_dates['date'].dt.day
df_dates['hora'] = df_dates['date'].dt.hour
df_dates['semana'] = df_dates['date'].dt.isocalendar().week
df_dates['is_ano_bisexto'] = df_dates['date'].dt.is_leap_year
df_dates['dia_semana'] = df_dates['date'].dt.weekday
df_dates['trimestre'] = df_dates['date'].dt.quarter
df_dates['dia_ano'] = df_dates['date'].dt.dayofyear
```
